## Abstract

Proglacial lakes across the Himalaya are expanding rapidly in both area and number, altering glacier dynamics and heightening the risk of glacier lake outburst floods (GLOFs). Yet, how individual glaciers respond dynamically to changes in lake level remains poorly constrained. Here, we first construct an idealized flowline model to examine how a lake-terminating glacier responds to lake-level perturbations under different sliding and calving laws, as well as to different bed topographies. We then repeat similar experiments using the observed geometry of Luggye Glacier, a lake-terminating glacier in the Lunana region of Bhutan. We find that, for a given calving law, mass delivery into the lake is enhanced by rising lake levels when water pressure influences the basal shear stress. In contrast, when basal shear stress is decoupled from water pressure, calving flux is dampened by rising water levels under two of the three calving laws considered. The choice of sliding and calving laws, therefore, are of primary importance in discerning the ice dynamical component of GLOF risk and glacier change.

## Plain Language Summary

Glaciers that end in lakes are retreating faster than glaciers without a lake, and across the Himalaya these lakes are growing, raising the risk of catastrophic floods for communities downstream. Using a numerical ice flow model, we tested how glaciers respond to small rises and drops in lake level. We find that conditions beneath the ice determine how the glacier reacts to changing lake levels. Therefore, applying the correct subglacial conditions to glacier models is of primary importance in understanding future glacier change and evaluating GLOF risk.

## 1 Introduction

Mountain glaciers have lost mass at an accelerating rate over recent decades (Hugonnet et al., 2021), while proglacial lakes have expanded simultaneously in number, area, and volume (Shugar et al., 2020). Both of these global trends are apparent across the Himalaya (Ahmed et al., 2021; Zhang et al., 2023), where the presence of proglacial lakes is associated with thinning, speedup, and retreat substantially above the regional mean (King et al., 2018). Many proglacial lakes, in addition to fundamentally altering dynamics at the terminal boundaries of glaciers, also pose an acute hazard by way of glacial lake outburst floods (GLOFs). GLOFs have caused widespread destruction and loss of life both globally and across the Himalaya (Taylor et al., 2023; Nie et al., 2018). For high-risk communities, better understanding the interplay between proglacial lake evolution and glacier dynamics is an immediate and high-priority concern (Wangchuk & Tsubaki, 2024; Mir et al., 2025).

The presence of a proglacial lake modifies glacier behaviour in three fundamental ways. First, proglacial water exerts compressive backpressure against the ice face, and so for a given terminus thickness, deeper water reduces extensional stress at the front (e.g., Cuffey & Paterson, 2010; van der Veen, 2013). Second, water pressure reduces the effective overburden pressure of the ice, consequently (though depending on the sliding law, as discussed later) reducing basal friction, thereby *increasing* extensional stresses near the front (Truffer et al., 2001; Tsai et al., 2015). Third, any proglacial water body introduces frontal mass loss processes, such as calving and subaqueous melt, absent at land-terminating glaciers, providing a mechanism for rapid retreat and enhancing mass wastage into the lake (Watson et al., 2020).

These mechanisms are represented in numerical glacier models through a combination of boundary conditions and constitutive relations. *Sliding laws* prescribe the relationship between basal shear stress and ice motion, and, in some formulations, couple this relationship to the effective pressure at the bed; *calving laws* parameterize the frontal mass loss processes described above, either by prescribing the rate at which the terminus migrates or by placing the calving front location directly. Meanwhile, the backpressure of the first mechanism enters the momentum balance as a boundary condition. Bed geometry acts as a further, orthogonal control on the whole system: the underlying topography sets both the local ice thickness at any point along the flowline and, near the terminus, the water depth encountered by an advancing or retreating front.

How these processes interact is not straightforward, and a glacier's net response to changing lake levels likely depends on the nature of the calving laws and sliding laws simultaneously at work. For instance, power-law sliding relations (Weertman, 1957) traditionally carry no dependence on water pressure, so a rising lake affects a glacier only through the first and third mechanisms above. In contrast, sliding laws invoking Coulomb failure (Tsai et al., 2015) allow basal resistance to decrease as a glacier's front approaches flotation. In this case, a rising lake level introduces competition between the increased backpressure against the terminus, which slows a glacier down, and the decreased basal resistance, which speeds it up. The calving law introduces a further level of complexity: two calving laws, even if framed in terms of the same underlying forcing, might produce fundamentally different terminus responses. Under a rate-based law, calving competes with ice velocity: a sufficiently fast-flowing glacier can maintain a steady terminus position even under strong calving forcing, with the retreat and advance rates in dynamic balance. Under a position-based law, the terminus is placed directly by the calving criterion, and no amount of forward flux can sustain a position beyond the prescribed threshold. In either case, since the calving law determines the location — and thereby the ice thickness and water depth — at the frontal boundary, the system is internally coupled in a way that makes the net response to a lake-level change difficult to predict without numerical modeling.

In this paper, we use the ice dynamics package *icepack* (Shapero et al., 2021) to systematically investigate the dynamic response of glaciers to lake-level perturbations, with attention to how the response varies depending on the types of sliding laws and calving laws used, as well as different underlying bed shapes. We perform experiments in an idealized setting, as well as on an observation-informed flowline model of Luggye Glacier, a lake-terminating glacier in Lunana, Bhutan.

### 1.1 Luggye Glacier

The Lunana region of northern Bhutan is home to four large lake-terminating glaciers: Luggye, Thorthormi, Raphstreng, and Bechung. At least 18 GLOFs have been recorded across Bhutan since the 1950s (Rinzin et al., 2023), and the Lunana complex has been the site of some of the most destructive. The most serious was the partial breach of Luggye Tsho, the proglacial lake at the terminus of Luggye Glacier, in October 1994. Floodwaters and debris surged approximately 90 km downstream, killing 21 people and causing extensive damage to cropland and infrastructure (Watanabe & Rothacher, 1996; Maurer et al., 2020). The lake has expanded overall since at least the 1970s, driven by (and likely driving) ongoing glacier retreat, and it continues to grow (Maurer et al., 2020). In response to the regional hazard, engineered interventions have been undertaken to lower lake levels, most notably a project targeting Thorthormi Tsho that succeeded in lowering its water level by approximately 5 m, though the lake remains dangerous (Rinzin et al., 2023). Luggye Tsho is currently assessed as among the most susceptible lakes in the Lunana complex to future GLOFs (Rinzin et al., 2023), making Luggye Glacier a natural focus for process-based modelling of glacier-lake dynamics.

## 2 Model and methods

We systematically investigate how the dynamic response of lake-terminating glaciers to lake-level perturbations depends on the choice of sliding law and calving law. We conduct three sets of flowline modeling experiments, each corresponding to an initial geometric configuration. The first configuration represents an idealized glacier whose terminus lies on a locally retrograde slope; the second is similar to the first but on a prograde slope. The third and final setup uses an observation-informed flowline representation of Luggye Glacier. In each case, we obtain a reference glacier geometry, either through spin-up (for the idealized configurations) or through bed inversion (for Luggye Glacier), for every sliding-calving law combination (see Section 2.4 for details). We then induce a series of perturbations to the lake level, ranging from $-10$ m to $+10$ m in integer increments, and we forward model each perturbed configuration for one year of simulated time. This yields 378 experiments in total (three geometric configurations $\times$ two sliding laws $\times$ three calving laws $\times$ twenty-one lake-level perturbations), each of which explores the dynamic and geometric responses to a given waterline perturbation under a particular sliding-calving law pair. At the end of each experiment, we record the changes to terminus position, calving flux, and terminus speed relative to each initial configuration's zero-perturbation baseline. The subsections below describe the model physics, sliding laws, calving laws, and initial configurations used in each case.

### 2.1 Ice flow approximation

Glacier flow is commonly approximated using either the Shallow Ice Approximation (SIA; Hutter, 1983), in which flow is primarily resisted by basal drag, or the Shallow Shelf Approximation (SSA; Morland, 1987; MacAyeal, 1989), in which longitudinal stress gradients dominate. For a lake-terminating glacier, neither alone is sufficient: basal drag likely dominates upstream, but it may decrease substantially if the terminus advances into deep water. We therefore use the Higher Order (HO) Approximation (Blatter, 1995), which captures both regimes and is appropriate across the range of conditions our experiments explore.

### 2.2 Sliding laws

Sliding laws can be categorized between two end-member forms. In pure power-law relations (Weertman, 1957), the basal shear stress $\tau_b$ depends only on the basal sliding velocity $u_b$, with no sensitivity to basal water pressure. This produces a relationship of the form

$$\tau_b = C\,|u_b|^{1/m}, \tag{1}$$

where $C$ and $m$ are constants. This is the first sliding law considered in our analysis. In contrast, in pure Coulomb laws (e.g., Truffer et al., 2001; Tsai et al., 2015), $\tau_b$ depends only on the effective pressure at the bed, so that basal resistance collapses to zero as ice approaches flotation. As our second sliding law, we implement an intermediate relation which retains the form of Equation 1 but includes effective pressure dependence:

$$\tau_b = C\,f_{af}\,|u_b|^{1/m}. \tag{2}$$

Here, $0 \le f_{af} \le 1$ is the clamped fraction above flotation, defined at each point along the flowline as a function of the local ice thickness and the local water depth. It equals unity where ice is well grounded, and decreases toward zero as the front approaches flotation. Consequently, where $f_{af} = 0$, the basal shear vanishes, as in a Coulomb law; and where $f_{af} = 1$, Equation 2 reduces to the power-law relation of Equation 1. Relations of this form are well-established, if somewhat *ad hoc* (e.g., Budd et al., 1979, 1984; Pattyn, 2002), and the formulation has a useful property for our purposes: the two laws are identical where ice is well-grounded, and they diverge only as the terminus approaches flotation. The effect of the sliding law choice is, therefore, concentrated precisely where lake-level changes act.

> *Implementation note.* In the code these are the `power_law` (Eq 1) and `effective_pressure` (Eq 2) sliding laws. Both share the friction coefficient `C`; `effective_pressure` multiplies it by the fraction above flotation $f_{af}$ (clamped to $[0.01, 1]$ for numerical stability), so the two coincide where ice is well grounded and diverge only near the terminus.

### 2.3 Calving laws

Likewise, calving laws can be separated conceptually into two categories. *Rate laws* prescribe the calving front migration as a per-unit-time length change; the terminus location then emerges from a dynamic balance between forward ice flow and calving. *Position laws* instead prescribe the calving front location directly, typically as a function of a diagnostic variable, so that the terminus is placed regardless of the underlying flow (Amaral et al., 2020). Because it is usually possible to observe terminus retreat only from above the waterline, we make no attempt to independently parameterize subaqueous melt. Thus, the "calving laws" discussed in this section implicitly describe the combination of mechanical calving and subaqueous melt. We first consider a depth-based rate law

$$u_c = k\,d_t, \tag{3}$$

with $u_c$ the calving rate, $k$ a scaling factor, and $d_t$ the water depth at the terminus. This constitutes a simplified version of the minimal calving law described by Oerlemans and Nick (2005), where we have divided the original formulation by frontal area to express the result in terms of length change, rather than calving flux. This calving law is used, for example, in the Open Global Glacier Model, OGGM (Recinos et al., 2019).

We also consider a simple position law, which sets the glacier length, $L$, to

$$L = \max\{x : f_{af} \geq f^*\}, \tag{4}$$

where $x$ is the along-flow coordinate and $f^* \geq 0$ is a tunable fraction-above-flotation threshold. That is, the front of the glacier retreats to the most downstream point at which it attains the prescribed degree of grounding (Vieli et al., 2001).

Both the position law and the rate law are ultimately forced by the same variable, water depth: it directly sets $d_t$ in Equation 3, and enters $f_{af}$ in Equation 4 through the local flotation criterion. This shared forcing allows us to better isolate the effect of the rate vs. position formulation rather than the choice of underlying forcing.

Finally, we also describe the trivial calving law,

$$u_t + u_c = 0, \tag{5}$$

in which the terminus velocity $u_t$ and the calving rate $u_c$ exactly cancel for a net zero rate of length change. We include this third calving law for two reasons. First, it is so commonly applied that it warrants inclusion for completeness: Equation 5 is implicitly used in any ice sheet or glacier model that does not include a mechanism for advance or retreat. Second, and more specifically to our experimental design, the trivial calving law has no free parameters, making it the natural choice for spinning up the reference configurations from which the rate and position laws are calibrated (Section 2.4).

> *Implementation note.* In the code these are `depth_rate` (Eq 3), `flotation` (Eq 4), and `trivial` (Eq 5). Equation 4 uses the non-strict inequality $f_{af} \geq f^*$ deliberately: since a floating terminus has $f_{af} = 0$ exactly, a strict inequality would prevent $f^* = 0$ from ever being satisfied at a floating face. Independently of the active calving law, every step also locates the grounding line itself (the most-downstream point with $f_{af}$ strictly positive) and reports the ice flux crossing it as `grounding_line_flux` — separating calving from front-only calving whenever a floating tongue is present.

### 2.4 Initial geometry and spinup

#### 2.4.1 Idealized flowline models

For the idealized experiments, we initialize the glacier as uniformly 200 m thick, and resting on an undulating bed. The average basal slope is a 5% prograde inclination, but the undulations create local deviations from this mean. In the first setup, the undulations are arranged so that the terminus rests on a locally retrograde slope; in the second, the terminus rests on a locally prograde slope. We use this undulating setup rather than imposing net prograde and net retrograde slopes on the basis that (1) by keeping the net basal slope the same in both cases, we permit both the prograde and retrograde configurations to closely coincide, thereby avoiding comparing dynamic responses of unlike geometries, and (2) we bypass the fundamental instabilities that arise as a result of flowline models on retrograde slopes (Weertman, 1974). In each case, we set the lake depth at the terminus to 50 m. We spin each configuration up to steady state under the trivial calving law of Equation 5 for each of the two sliding laws, yielding four distinct reference profiles.

For all four steady-state reference profiles, we record the terminus ice speed $u_t$ and the grounding line water depth $d_t$, and we use these to calibrate the remaining two calving laws so that each is consistent with the spun-up geometry. For the rate law, we use the fact that $u_t = u_c$ in each reference configuration, and we set $k = u_t / d_t$. Likewise, for the position law, we set $f^*$ equal to the observed fraction above flotation at either reference terminus. With these choices, the calibrated rate and position laws are already in equilibrium with the reference geometries, and no further spinup is required. Now for a given bed shape and sliding law, we have a single spun-up geometry in equilibrium with three distinct calving laws, providing a common starting point from which to induce perturbations.

#### 2.4.2 Luggye flowline model

For the model of Luggye Glacier, flowline surface elevations are obtained from a drone-derived DEM, and basal topography is recovered using the bed inversion technique of van Pelt et al. (2013). Starting from an initial bed elevation guess, we forward-model the ice flow for one year and compare the resulting change in surface elevation. Reasoning that the surface elevation profile should not significantly change in one year of simulated time, we apply a correction to the bed in the direction opposite the surface misfit, and repeat. This procedure is iterated until the simulated surface remains stable over the one-year simulation, at which point the bed elevation is considered converged. We perform this inversion once per sliding law, using the trivial calving law to hold the terminus in place, yielding two distinct beds.

As Luggye Glacier is likely not in true steady state, we do not spin up further following the bed inversion. Instead, we immediately apply the same calibration procedure as for the idealized case: at either post-inversion terminus, we record $u_t$ and $d_t$, set $k = u_t / d_t$ for the rate law, and set $f^*$ equal to the resulting fraction above flotation at the terminus for the position law. This ensures that all three calving laws are instantaneously consistent with the same post-inversion geometry for a given sliding law: each calving law predicts zero net terminus migration at $t = 0$. Unlike with the steady-state idealized case, however, some drift in terminus position is to be expected over time, even at zero perturbation. For this reason, all diagnostics in Section 3 are reported relative to the zero-perturbation run for each configuration, rather than against an absolute reference.

```python
# Result cache
cache_dir = f'{data_path}/cache'
os.makedirs(cache_dir, exist_ok=True)


def _fingerprint_norm(v):
    if isinstance(v, Constant):
        v = float(v)
    if isinstance(v, np.ndarray):
        return [round(float(x), 12) for x in v.ravel()]
    if isinstance(v, (bool, np.bool_)):
        return bool(v)
    if isinstance(v, (int, np.integer)):
        return int(v)
    if isinstance(v, (float, np.floating)):
        return round(float(v), 12)
    if isinstance(v, dict):
        return {str(k): _fingerprint_norm(x) for k, x in sorted(v.items(), key=lambda kv: str(kv[0]))}
    if isinstance(v, (list, tuple, set)):
        seq = sorted(v, key=str) if isinstance(v, set) else v
        return [_fingerprint_norm(x) for x in seq]
    return str(v)


def fingerprint(**parts):
    blob = json.dumps(_fingerprint_norm(parts), sort_keys=True)
    return hashlib.sha1(blob.encode()).hexdigest()[:16]


def experiment_time_controls(tag):
    if tag == 'luggye':
        return dict(
            ramp_time = ramp_time if luggye_ramp_time is None else luggye_ramp_time,
            ramp_Δt   = ramp_Δt   if luggye_ramp_Δt   is None else luggye_ramp_Δt,
            hold_time = hold_time if luggye_hold_time is None else luggye_hold_time,
            hold_Δt   = hold_Δt   if luggye_hold_Δt   is None else luggye_hold_Δt)
    return dict(ramp_time=ramp_time, ramp_Δt=ramp_Δt,
                hold_time=hold_time, hold_Δt=hold_Δt)


def physics_parts():
    return dict(A=A, C=C, u_0=u_0, min_thickness=min_thickness,
                baseline_waterline=baseline_waterline,
                extra_length=extra_length,
                sliding_laws=sliding_laws, calving_laws=calving_laws,
                reference_sliding_law=reference_sliding_law)


def force_for(flag, tag):
    if isinstance(flag, bool):
        return flag
    if isinstance(flag, str):
        return flag == tag
    return tag in set(flag)


def cache_file(kind, key, ext):
    return f'{cache_dir}/{kind}_{key}.{ext}'


def cache_hit(kind, key, ext, force=False):
    if force or not use_cache:
        return None
    p = cache_file(kind, key, ext)
    return p if os.path.exists(p) else None


def save_state_cache(kind, key, state):
    mesh = state['mesh']
    x = np.sort(np.asarray(mesh.X, dtype=float))
    H = np.asarray(icepack.depth_average(state['H']).at(x, tolerance=1e-10), dtype=float)
    np.savez(cache_file(kind, key, 'npz'), x=x, H=H,
             L=float(mesh.length), n_cells=int(n_cells))


def load_state_cache(p, sliding_law, direction):
    d = np.load(p)
    if not np.isclose(float(d['L']), float(length)) or int(d['n_cells']) != int(n_cells):
        return None
    st = make_initial_state(sliding_law, direction=direction)
    Q = st['H'].function_space()
    st['H'] = function_from_xprofile(Q, st['mesh'], d['H'], x_data=d['x'])
    st['s'] = compute_surface_with_waterline(thickness=st['H'], bed=st['bed'],
                                             waterline=baseline_waterline)
    solver = make_flow_solver(waterline=baseline_waterline,
                              water_aware_friction=(sliding_law == 'effective_pressure'))
    st['u'] = solver.diagnostic_solve(velocity=st['u'], thickness=st['H'],
                                      surface=st['s'], fluidity=A, friction=C)
    return st


# Round-off levels applied before hashing baselines, so sub-tolerance
# thickness or calving-parameter noise never re-fingerprints an identical run.
_FP_H_DECIMALS = 3
_FP_PARAM_DECIMALS = 6
_FP_GRID_POINTS = 512


def baselines_parts(baselines):
    parts = {}
    for (sl, direction, cl), b in sorted(baselines.items(), key=lambda kv: str(kv[0])):
        st = b['state']
        L = float(st['mesh'].length)
        xg = np.linspace(0.0, L, _FP_GRID_POINTS)
        Hg = np.asarray(icepack.depth_average(st['H']).at(xg, tolerance=1e-10),
                        dtype=float)
        parts[f'{sl}|{direction}|{cl}'] = dict(
            L=round(L, _FP_H_DECIMALS),
            H=[round(float(v), _FP_H_DECIMALS) for v in Hg],
            depth_rate=round(float(b['depth_rate']), _FP_PARAM_DECIMALS),
            f_star=round(float(b['f_star']), _FP_PARAM_DECIMALS))
    return parts
```

```python
# Model setup: mesh, initial state, flow solver, diagnostics
def compute_surface_with_waterline(thickness, bed, waterline):
    """Surface elevation consistent with the given waterline: s = max(bed + H, waterline + (1 - ϱ) H).
    Required whenever the waterline is perturbed (icepack.compute_surface assumes waterline = 0)."""
    ϱ = icepack.constants.ice_density / icepack.constants.water_density
    Q = thickness.function_space()
    return Function(Q).interpolate(
        firedrake.max_value(bed + thickness, waterline + (1 - ϱ) * thickness))


def make_mesh():
    mesh = centerline_mesh(n_cells=n_cells, length=length)
    V = FunctionSpace(mesh.mesh, 'CG', 1, vfamily='GL', vdegree=2)
    Q = FunctionSpace(mesh.mesh, 'CG', 1, vfamily='R',  vdegree=0)
    return mesh, V, Q


def _interp_extrap(xq, xd, vd):
    # np.interp clamps to edge values outside the data range: a neutral seaward
    # extension for the advance room (a constant-depth lake), rather than
    # continuing the terminal bed slope, which for Luggye's up-tilted terminus
    # would invent proglacial land the inversion never saw.
    xq = np.asarray(xq, float)
    return np.interp(xq, xd, vd)


def function_from_xprofile(Q, mesh, v_data, x_data=None):
    if x_data is None:
        x_data = geom_X
    xfun = Function(Q).interpolate(SpatialCoordinate(mesh.mesh)[0])
    f = Function(Q)
    f.dat.data[:] = _interp_extrap(xfun.dat.data_ro, x_data, v_data)
    return f


def make_initial_state(sliding_law=None, direction='none'):
    mesh, V, Q = make_mesh()
    sl = sliding_law if sliding_law is not None else reference_sliding_law
    key = (sl, direction)
    if key not in geom_bed:
        key = (reference_sliding_law, direction)
    bed = function_from_xprofile(Q, mesh, geom_bed[key])
    H   = function_from_xprofile(Q, mesh, geom_thickness[key])
    H   = Function(Q).interpolate(max_value(H, Constant(min_thickness)))
    s   = compute_surface_with_waterline(thickness=H, bed=bed, waterline=baseline_waterline)
    _, ζ = SpatialCoordinate(mesh.mesh)
    u_shape = (5/4) * (1 - (1 - ζ)**4)     # Glen n=3 laminar, ∫dζ = 1
    u = Function(V).project(u_0 * u_shape)
    return {'mesh': mesh, 'bed': bed, 'H': H, 's': s, 'u': u}


def make_flow_solver(waterline=baseline_waterline, water_aware_friction=False):
    """HO flow solver carrying the chosen sliding law and the lake back-pressure
    terminus boundary condition. water_aware_friction=True selects Eq 2; False selects Eq 1."""
    def basal_friction(**kwargs):
        H = kwargs['thickness']; s = kwargs['surface']
        u = kwargs['velocity'];  C_ = kwargs['friction']
        ϱ = icepack.constants.ice_density / icepack.constants.water_density
        h_af = s - (waterline + (1 - ϱ) * H)
        frac_af = (firedrake.min_value(firedrake.max_value(h_af / H, 0.01), 1)
                   if water_aware_friction else 1)
        return icepack.models.hybrid.bed_friction(velocity=u, friction=C_ * frac_af)

    def terminus(**kwargs):
        H = kwargs['thickness']; s = kwargs['surface']; u = kwargs['velocity']
        mesh = u.ufl_domain()
        _, ζ = SpatialCoordinate(mesh)
        ν = FacetNormal(mesh)
        ρ_I = icepack.constants.ice_density
        ρ_W = icepack.constants.water_density
        G = icepack.constants.gravity
        base = s - H
        ζ_w = firedrake.min_value(firedrake.max_value((waterline - base) / H, 0), 1)
        p_I = ρ_I * G * H * (1 - ζ)
        p_W = ρ_W * G * H * firedrake.max_value(ζ_w - ζ, 0)
        return (p_I - p_W) * u * ν[0] * H

    flow_model = icepack.models.HybridModel(friction=basal_friction, terminus=terminus)
    return icepack.solvers.FlowSolver(flow_model, dirichlet_ids=[1])


def terminus_x(state):
    return float(state['mesh'].length)


def diagnostics(state, waterline=baseline_waterline):
    mesh = state['mesh']
    H, s, u = state['H'], state['s'], state['u']
    base = Function(H.function_space()).project(s - H)
    x_t = terminus_x(state)
    H_t = float(H.at(x_t, 0, tolerance=1e-10))
    base_t = float(base.at(x_t, 0, tolerance=1e-10))
    d_t = max(waterline - base_t, 0)
    u_t = float(icepack.depth_average(u).at(x_t, tolerance=1e-10))
    H_bar = np.array(icepack.depth_average(H).at(mesh.X, tolerance=1e-10))
    x_arr = np.asarray(mesh.X, dtype=float)
    order = np.argsort(x_arr)
    V = float(np.trapezoid(H_bar[order], x_arr[order]))
    return {'L': x_t, 'H_t': H_t, 'base_t': base_t,
            'd_t': d_t, 'u_t': u_t, 'V': V}
```

```python
# Calving-law factory and per-baseline calibration
GROUNDED_EPS = 1e-6   # f_af must exceed this to count as genuinely grounded


def fraction_above_flotation_profile(state, waterline):
    """(x_sorted, frac_af_sorted) at every mesh node."""
    mesh = state['mesh']
    H, s = state['H'], state['s']
    x = np.asarray(mesh.X, dtype=float)
    order = np.argsort(x)
    x = x[order]
    ϱ = icepack.constants.ice_density / icepack.constants.water_density
    h_af = Function(H.function_space()).project(s - (waterline + (1 - ϱ) * H))
    H_s   = np.asarray(icepack.depth_average(H).at(x, tolerance=1e-10), dtype=float)
    haf_s = np.asarray(icepack.depth_average(h_af).at(x, tolerance=1e-10), dtype=float)
    with np.errstate(divide='ignore', invalid='ignore'):
        frac = np.where(H_s > 0, haf_s / H_s, 0.0)
    return x, frac


def _threshold_crossing(x, frac, threshold, L_terminus, all_satisfied_value):
    """Most-inland point of the near-terminus zone with frac >= threshold,
    linearly interpolated. Returns all_satisfied_value if the terminus already satisfies it."""
    g = frac - threshold
    sat = g >= 0.0
    n = len(x)
    if sat[-1]:
        return all_satisfied_value
    if not np.any(sat):
        return float(x[0])
    i = n - 1
    while i > 0 and not sat[i - 1]:
        i -= 1
    g_lo, g_hi = float(g[i - 1]), float(g[i])
    if g_lo == g_hi:
        return float(x[i - 1])
    frac_root = g_lo / (g_lo - g_hi)
    return min(float(x[i - 1]) + frac_root * float(x[i] - x[i - 1]), L_terminus)


def grounding_line_position(state, waterline):
    """Most-downstream x with f_af > GROUNDED_EPS, capped at the terminus."""
    x, frac = fraction_above_flotation_profile(state, waterline)
    L_t = terminus_x(state)
    return _threshold_crossing(x, frac, GROUNDED_EPS, L_t, all_satisfied_value=L_t)


def grounding_line_depth(state, waterline):
    """d_gl = max(waterline - base(x_gl), 0); the depth Eq. 3 sees when a floating tongue is present."""
    x_gl = grounding_line_position(state, waterline)
    H, s = state['H'], state['s']
    base = Function(H.function_space()).project(s - H)
    base_gl = float(base.at(x_gl, 0, tolerance=1e-10))
    return max(waterline - base_gl, 0.0), x_gl


def make_calving_law(name, depth_rate=None, f_star=None):
    if name == 'depth_rate':           # Eq. 3: c = k · d_gl
        k = (depth_rate if depth_rate is not None
             else per_sliding(depth_rate_default, reference_sliding_law))
        return lambda d_gl, u_t: k * d_gl

    if name == 'trivial':              # Eq. 5: held by the stepper
        return lambda d_gl, u_t: u_t

    if name == 'flotation':            # Eq. 4: L = max{x : f_af >= f*}
        fs = (f_star if f_star is not None
              else per_sliding(f_star_default, reference_sliding_law))
        if fs <= 1e-9:
            return lambda state, waterline: float('inf')
        def flotation_position(state, waterline):
            x, frac = fraction_above_flotation_profile(state, waterline)
            L_t = terminus_x(state)
            return _threshold_crossing(x, frac, fs, L_t,
                                        all_satisfied_value=float('inf'))
        return flotation_position

    raise ValueError(name)


def calibrate_depth_rate(state, waterline):
    """k = u_t / d_gl at the reference state (Eq. 3)."""
    d_gl, _ = grounding_line_depth(state, waterline)
    if d_gl <= 1e-6:
        return None
    mesh = state['mesh']
    u_t  = float(icepack.depth_average(state['u']).at(mesh.length, tolerance=1e-10))
    return u_t / d_gl


def calibrate_f_star(state, waterline):
    """f* = fraction above flotation at the terminus (Eq. 4)."""
    mesh = state['mesh']
    H, s = state['H'], state['s']
    ϱ = icepack.constants.ice_density / icepack.constants.water_density
    h_af = Function(H.function_space()).project(s - (waterline + (1 - ϱ) * H))
    x_t = mesh.length
    H_t   = float(icepack.depth_average(H).at(x_t, tolerance=1e-10))
    haf_t = float(icepack.depth_average(h_af).at(x_t, tolerance=1e-10))
    return haf_t / H_t if H_t > 0 else None
```

```python
# State copy, mesh rebuild, and one timestep
def copy_state(state):
    return {'mesh': state['mesh'],
            'bed': state['bed'].copy(deepcopy=True),
            'H':   state['H'].copy(deepcopy=True),
            's':   state['s'].copy(deepcopy=True),
            'u':   state['u'].copy(deepcopy=True)}


def _build_state_on_mesh(mesh_target, bed_0, H_src, u_src, *, waterline,
                         ice_free_H=None, ice_free_u=None):
    """Extend (H, bed, s, u) onto mesh_target; surface rebuilt against the current waterline."""
    H_new = (extend_to_mesh(function=H_src, mesh=mesh_target, ice_free_value=ice_free_H)
             if ice_free_H is not None else
             extend_to_mesh(function=H_src, mesh=mesh_target))
    bed_new = extend_to_mesh(function=bed_0, mesh=mesh_target)
    s_new = compute_surface_with_waterline(thickness=H_new, bed=bed_new, waterline=waterline)
    u_new = (extend_to_mesh(function=u_src, mesh=mesh_target, ice_free_value=ice_free_u)
             if ice_free_u is not None else
             extend_to_mesh(function=u_src, mesh=mesh_target))
    return {'mesh': mesh_target, 'bed': bed_new, 'H': H_new, 's': s_new, 'u': u_new}


def step_with_remeshing(state, bed_0, waterline, water_aware_friction,
                        calving_law_name, calving_law_fn, Δt_step,
                        L_domain):
    """Advance one Δt. Returns (new_state, event_dict).

    Calving-law dispatch (Section 2.3):
        trivial    — mesh held fixed at L_0 (Eq. 5); c = u_t as diagnostic.
        rate-based — net Δx = (u_t − c) Δt with c = k · d_gl (Eq. 3).
        position   — advance by u_t Δt, then calve back to L = max{x : f_af >= f*} (Eq. 4) when inland.

    Grounding line is computed independently of the calving law (used to separate
    calving flux from mass actually delivered to the lake).
    """
    mesh_1 = state['mesh']
    H, u, bed = state['H'], state['u'], state['bed']
    L_0 = terminus_x(state)

    s = compute_surface_with_waterline(thickness=H, bed=bed, waterline=waterline)
    flow_solver = make_flow_solver(waterline=waterline,
                                   water_aware_friction=water_aware_friction)
    u = flow_solver.diagnostic_solve(velocity=u, thickness=H, surface=s,
                                     fluidity=A, friction=C)
    H = flow_solver.prognostic_solve(Δt_step, thickness=H, velocity=u,
                                     thickness_inflow=H_initial,
                                     accumulation=Constant(0))
    s = compute_surface_with_waterline(thickness=H, bed=bed, waterline=waterline)

    Q = H.function_space()
    base = Function(Q).project(s - H)
    H_t = float(H.at(L_0, 0, tolerance=1e-10))
    base_t = float(base.at(L_0, 0, tolerance=1e-10))
    d_t = max(waterline - base_t, 0)
    u_t = float(icepack.depth_average(u).at(L_0, tolerance=1e-10))

    state_phys = {'mesh': mesh_1, 'bed': bed, 'H': H, 's': s, 'u': u}
    _, frac_af = fraction_above_flotation_profile(state_phys, waterline)
    f_af_terminus = float(frac_af[-1])

    x_gl    = grounding_line_position(state_phys, waterline)
    base_gl = float(base.at(x_gl, 0, tolerance=1e-10))
    d_gl    = max(waterline - base_gl, 0.0)
    H_gl    = float(icepack.depth_average(H).at(x_gl, tolerance=1e-10))
    u_gl    = float(icepack.depth_average(u).at(x_gl, tolerance=1e-10))
    grounding_line_volume = u_gl * H_gl * Δt_step

    # An advancing front can move at most u_t·Δt AND never beyond L_domain
    # (the extended master bed), which is the extent of validly-defined geometry.
    L_adv = min(L_0 + u_t * Δt_step, L_domain)

    c_inst = None
    L_proposed = None
    if calving_law_name == 'trivial':
        state_out = state_phys
        L_new = L_0

    elif calving_law_name in RATE_BASED_LAWS:
        c_inst = float(calving_law_fn(d_gl, u_t))
        Δx = (u_t - c_inst) * Δt_step
        mesh_2 = centerline_mesh(mesh=mesh_1, n_cells=n_cells, extra_length=Δx)
        state_out = _build_state_on_mesh(mesh_2, bed_0, H, u, waterline=waterline,
                                         ice_free_H=H_t, ice_free_u=u_t)
        L_new = float(mesh_2.length)

    elif calving_law_name in POSITION_BASED_LAWS:
        L_proposed = float(calving_law_fn(state_phys, waterline))
        Δx = min(L_adv, L_proposed) - L_0
        if Δx == 0.0:
            state_out = state_phys
        else:
            mesh_2 = centerline_mesh(mesh=mesh_1, n_cells=n_cells, extra_length=Δx)
            state_out = _build_state_on_mesh(mesh_2, bed_0, H, u, waterline=waterline,
                                             ice_free_H=H_t, ice_free_u=u_t)
        L_new = terminus_x(state_out)

    else:
        raise ValueError(f'Unknown calving law: {calving_law_name}')

    # Grounding line on the output state — the honest partner for L_new (same mesh, same waterline).
    x_gl_out = grounding_line_position(state_out, waterline)

    event = {'dL': L_new - L_0, 'd_t': d_t, 'd_gl': d_gl,
             'u_t': u_t, 'H_t': H_t,
             'x_gl': x_gl, 'x_gl_out': x_gl_out, 'u_gl': u_gl, 'H_gl': H_gl,
             'grounding_line_volume': grounding_line_volume,
             'L_0': L_0, 'L_new': L_new, 'L_adv': L_adv,
             'L_proposed': L_proposed, 'c_inst': c_inst,
             'f_af_terminus': f_af_terminus,
             'Δt': Δt_step}
    return state_out, event
```

```python
# Spin-up and experiment drivers
def prepare_extended_bed(state, sliding_law=None, direction='none'):
    """Master bed on a mesh extended seaward by extra_length (advance room)."""
    mesh_1 = state['mesh']
    mesh_0 = centerline_mesh(mesh=mesh_1, n_cells=n_cells, extra_length=extra_length)
    Q0 = FunctionSpace(mesh_0.mesh, 'CG', 1, vfamily='R', vdegree=0)
    sl = sliding_law if sliding_law is not None else reference_sliding_law
    key = (sl, direction) if (sl, direction) in geom_bed else (reference_sliding_law, direction)
    bed_0 = function_from_xprofile(Q0, mesh_0, geom_bed[key])
    return mesh_0, bed_0


def run_spinup(initial_state, bed_0, sliding_law, calving_law, *,
               max_time, Δt_init, Δt_final, Δt_ramp, L_domain,
               tol_V_frac=1e-5, tol_L_frac=1e-5,
               desc='', depth_rate=None, f_star=None):
    """Integrate until steady state (fractional rates in V and L below
    tolerance for two consecutive steps) or max_time. Δt ramps linearly
    from Δt_init to Δt_final over Δt_ramp years."""
    def timestep(time):
        frac = min(time / Δt_ramp, 1.0) if Δt_ramp > 0 else 1.0
        return Δt_init + frac * (Δt_final - Δt_init)

    state = copy_state(initial_state)
    water_aware = (sliding_law == 'effective_pressure')
    calving_fn = make_calving_law(calving_law, depth_rate=depth_rate, f_star=f_star)
    time = 0.0
    prev_L, prev_V = None, None
    converged_at = None
    convergence_count = 0
    dV_frac_dt, dL_frac_dt = np.inf, np.inf
    pbar = tqdm(total=max_time, desc=desc, unit='yr', leave=False, dynamic_ncols=True)
    while time < max_time:
        Δt_step = min(timestep(time), max_time - time)
        if Δt_step <= 1e-12 * max_time:
            break
        state, event = step_with_remeshing(
            state, bed_0=bed_0, waterline=baseline_waterline,
            water_aware_friction=water_aware,
            calving_law_name=calving_law, calving_law_fn=calving_fn,
            Δt_step=Δt_step, L_domain=L_domain)
        time += Δt_step
        diag = diagnostics(state)
        L, V = diag['L'], diag['V']
        if prev_L is not None:
            dV_frac_dt = (V - prev_V) / max(abs(prev_V), 1.0) / Δt_step
            dL_frac_dt = (L - prev_L) / max(abs(prev_L), 1.0) / Δt_step
            if abs(dV_frac_dt) < tol_V_frac and abs(dL_frac_dt) < tol_L_frac:
                convergence_count += 1
                if convergence_count >= 2:
                    converged_at = time
            else:
                convergence_count = 0
        pbar.update(Δt_step)
        pbar.set_postfix({'t': f'{time:.0f}', 'L': f'{L:.0f}', 'u': f'{event["u_t"]:.1f}',
                          'dV/V/dt': f'{dV_frac_dt:+.1e}', 'dL/L/dt': f'{dL_frac_dt:+.1e}'})
        if converged_at is not None:
            break
        prev_L, prev_V = L, V
    pbar.close()
    status = (f'converged at t = {converged_at:.0f} yr' if converged_at is not None
              else f'reached max_time = {max_time:.0f} yr WITHOUT converging')
    print(f'    [{desc}] {status}  (dV/V/dt = {dV_frac_dt:+.1e}/yr, dL/L/dt = {dL_frac_dt:+.1e}/yr)')
    return state


def _waterline_at(time_end_of_step, offset, ramp_duration=None):
    """Waterline at end of a step: linear ramp from baseline to baseline+offset over [0, ramp_duration], held at target thereafter."""
    rd = ramp_time if ramp_duration is None else ramp_duration
    if rd <= 0:
        return baseline_waterline + offset
    frac = min(time_end_of_step / rd, 1.0)
    return baseline_waterline + frac * offset


def _dump_state_on_failure(state, waterline, Δt_step, context, exc):
    print()
    print('================ SOLVE FAILED — state dump ================')
    print(f'  context  : {context}')
    print(f'  exception: {type(exc).__name__}: {exc}')
    print(f'  waterline: {waterline:+.4f} m   '
          f'(baseline {baseline_waterline:+.4f}, applied offset {waterline - baseline_waterline:+.4f})')
    print(f'  Δt       : {Δt_step:.6e} yr')
    try:
        mesh = state['mesh']
        print(f'  L_0      : {float(mesh.length):.2f} m   (n_cells = {n_cells})')
    except Exception as e:
        print(f'  (mesh inspect failed: {e})')
    for name, key in [('H  ', 'H'), ('s  ', 's'), ('bed', 'bed'), ('u  ', 'u')]:
        try:
            arr = np.asarray(state[key].dat.data_ro)
            print(f'  {name}    : min={float(arr.min()):+12.4f}  '
                  f'max={float(arr.max()):+12.4f}  '
                  f'mean={float(arr.mean()):+12.4f}  '
                  f'NaN={int(np.isnan(arr).sum()):3d}  '
                  f'Inf={int(np.isinf(arr).sum()):3d}')
        except Exception as e:
            print(f'  ({name} stats failed: {e})')
    try:
        d = diagnostics(state, waterline=waterline)
        print(f'  terminus : H_t={d["H_t"]:.2f}  base_t={d["base_t"]:+.2f}  '
              f'd_t={d["d_t"]:.2f}  u_t={d["u_t"]:.2f}')
    except Exception as e:
        print(f'  (terminus diagnostics failed: {e})')
    try:
        _, frac_af = fraction_above_flotation_profile(state, waterline)
        d_gl, x_gl = grounding_line_depth(state, waterline)
        near_float_count = int(np.sum(frac_af < 0.05))
        print(f'  frac_af  : min={float(frac_af.min()):.4f}  '
              f'max={float(frac_af.max()):.4f}  '
              f'at-terminus={float(frac_af[-1]):.4f}  '
              f'(<0.05: {near_float_count}/{len(frac_af)} cells)')
        print(f'  GL       : x_gl={x_gl:.2f} m   d_gl={d_gl:.2f} m')
    except Exception as e:
        print(f'  (frac_af / GL diagnostics failed: {e})')
    print('============================================================')
    print()


def run_experiment(baselines, sliding_law, calving_law, offset, direction='none',
                   trace=None, time_controls=None, final_state=None):
    """Forward-model one (sliding, calving, offset) config through ramp+hold.
    Returns end-state diagnostics plus mean grounding-line flux rate."""
    tc = time_controls if time_controls is not None else dict(
        ramp_time=ramp_time, ramp_Δt=ramp_Δt, hold_time=hold_time, hold_Δt=hold_Δt)
    _ramp_time, _ramp_Δt = tc['ramp_time'], tc['ramp_Δt']
    _hold_time, _hold_Δt = tc['hold_time'], tc['hold_Δt']

    baseline = baselines[(sliding_law, direction, calving_law)]
    state = copy_state(baseline['state'])
    bed_0 = baseline['bed_0']
    L_domain = float(baseline['mesh_0'].length)
    water_aware = (sliding_law == 'effective_pressure')
    f_star_val = baseline.get('f_star')
    calving_fn = make_calving_law(calving_law,
                                  depth_rate=baseline.get('depth_rate'),
                                  f_star=f_star_val)
    cum_gl_volume = 0.0
    step_idx = 0
    elapsed = 0.0

    if trace is not None:
        _d0 = diagnostics(state, waterline=baseline_waterline)
        _, _frac0 = fraction_above_flotation_profile(state, baseline_waterline)
        trace.append({
            'step_idx': 0, 'phase': 'initial', 'time': 0.0,
            'waterline': baseline_waterline, 'Δt': 0.0,
            'L_0': _d0['L'], 'L_new': _d0['L'], 'L_adv': _d0['L'],
            'L_proposed': np.nan, 'Δx': 0.0,
            'u_t': _d0['u_t'], 'H_t': _d0['H_t'],
            'd_t': _d0['d_t'],
            'd_gl': grounding_line_depth(state, baseline_waterline)[0],
            'x_gl': float(grounding_line_position(state, baseline_waterline)),
            'x_gl_out': float(grounding_line_position(state, baseline_waterline)),
            'H_gl': np.nan, 'u_gl': np.nan,
            'gl_flux_rate': np.nan, 'cum_gl_volume': 0.0,
            'f_af_terminus': float(_frac0[-1]), 'f_star': f_star_val,
            'c_inst': np.nan,
        })

    def _step(wl, Δt, phase):
        nonlocal state, cum_gl_volume, step_idx, elapsed
        step_idx += 1
        try:
            state_new, event = step_with_remeshing(
                state, bed_0=bed_0, waterline=wl,
                water_aware_friction=water_aware,
                calving_law_name=calving_law, calving_law_fn=calving_fn,
                Δt_step=Δt, L_domain=L_domain)
        except Exception as e:
            ctx = (f'sliding={sliding_law}, direction={direction}, '
                   f'calving={calving_law}, '
                   f'offset={offset:+.2f} m; {phase} phase step #{step_idx}')
            _dump_state_on_failure(state, wl, Δt, ctx, e)
            raise
        state = state_new
        cum_gl_volume += event['grounding_line_volume']
        elapsed += Δt
        if trace is not None:
            trace.append({
                'step_idx': step_idx, 'phase': phase, 'time': elapsed,
                'waterline': wl, 'Δt': Δt,
                'L_0': event['L_0'], 'L_new': event['L_new'],
                'L_adv': event['L_adv'], 'L_proposed': event['L_proposed'],
                'Δx': event['dL'],
                'u_t': event['u_t'], 'H_t': event['H_t'],
                'd_t': event['d_t'], 'd_gl': event['d_gl'],
                'x_gl': event['x_gl'], 'x_gl_out': event['x_gl_out'],
                'H_gl': event['H_gl'], 'u_gl': event['u_gl'],
                'gl_flux_rate': event['grounding_line_volume'] / Δt if Δt > 0 else np.nan,
                'cum_gl_volume': cum_gl_volume,
                'f_af_terminus': event['f_af_terminus'], 'f_star': f_star_val,
                'c_inst': event['c_inst'],
            })

    wl_target = baseline_waterline + offset

    # Phase 1: linear ramp to target offset.
    if _ramp_time > 0:
        n_ramp = max(1, int(round(_ramp_time / _ramp_Δt)))
        Δt_ramp_exact = _ramp_time / n_ramp
        for k in range(n_ramp):
            time_end = (k + 1) * Δt_ramp_exact
            wl = _waterline_at(time_end, offset, ramp_duration=_ramp_time)
            _step(wl, Δt_ramp_exact, phase='ramp')

    # Phase 2: hold at target offset.
    if _hold_time > 0:
        n_hold = max(1, int(round(_hold_time / _hold_Δt)))
        Δt_hold_exact = _hold_time / n_hold
        for k in range(n_hold):
            _step(wl_target, Δt_hold_exact, phase='hold')

    total_time = _ramp_time + _hold_time
    d = diagnostics(state, waterline=wl_target)
    gl_pos_final = grounding_line_position(state, wl_target)

    if final_state is not None:
        final_state.update(state=state, bed_0=bed_0, mesh_0=baseline['mesh_0'])

    return {'sliding_law': sliding_law, 'direction': direction,
            'calving_law': calving_law,
            'waterline_offset': float(offset),
            'L': d['L'], 'u_t': d['u_t'],
            'gl_flux': cum_gl_volume / total_time,
            'gl_pos': float(gl_pos_final)}


def run_all_experiments(baselines, tag, configs=None):
    """Run the sweep, return one row per run. Cached under a fingerprint of
    the baselines, time controls and offset list; written to <tag>_response.csv."""
    if configs is None:
        configs = [(sl, 'none') for sl in sliding_laws]

    tc = experiment_time_controls(tag)

    key = fingerprint(kind='experiments', tag=tag, configs=configs,
                      offsets=waterline_offsets,
                      ramp_time=tc['ramp_time'], ramp_Δt=tc['ramp_Δt'],
                      hold_time=tc['hold_time'], hold_Δt=tc['hold_Δt'],
                      baselines=baselines_parts(baselines),
                      **physics_parts())
    hit = cache_hit('experiments', key, 'csv',
                    force=force_for(force_rerun_experiments, tag))

    final = None
    if hit is not None:
        cached = pd.read_csv(hit)
        if EXPERIMENT_COLUMNS.issubset(cached.columns):
            final = cached
            print(f'[{tag}] loaded cached sweep from {os.path.basename(hit)} '
                  f'({len(final)} runs). To redo it, set '
                  f'force_rerun_experiments = {tag!r}.')
        else:
            missing = sorted(EXPERIMENT_COLUMNS - set(cached.columns))
            print(f'[{tag}] cached sweep {os.path.basename(hit)} is missing '
                  f'{missing}; recomputing.')

    if final is None:
        rows = []
        for sl, direction in tqdm(configs, desc=f'{tag}: configs'):
            for cl in tqdm(calving_laws, desc=f'{tag}: {sl}/{direction}', leave=False):
                for off in tqdm(waterline_offsets,
                                desc=f'{tag}: {sl}/{direction}/{cl}', leave=False):
                    rows.append(run_experiment(baselines, sl, cl, off,
                                               direction=direction,
                                               time_controls=tc))
        final = pd.DataFrame(rows)
        final.to_csv(cache_file('experiments', key, 'csv'), index=False)

    final = add_differences(final)
    final.to_csv(f'{data_path}/{tag}_response.csv', index=False)
    return final


EXPERIMENT_COLUMNS = {'sliding_law', 'direction', 'calving_law',
                      'waterline_offset', 'L', 'u_t', 'gl_flux', 'gl_pos'}


def add_differences(final):
    """Add dL, du_t, dgl_flux, dgl_pos: each run minus the zero-offset run of the same baseline."""
    final = final.copy()
    for col in ('L', 'u_t', 'gl_flux', 'gl_pos'):
        final[f'd{col}'] = np.nan
    for (sl, direction, cl), grp in final.groupby(
            ['sliding_law', 'direction', 'calving_law']):
        zero = grp[np.isclose(grp['waterline_offset'], 0.0)]
        if not len(zero):
            continue
        ref = zero.iloc[0]
        m = ((final['sliding_law'] == sl) & (final['direction'] == direction) &
             (final['calving_law'] == cl))
        for col in ('L', 'u_t', 'gl_flux', 'gl_pos'):
            final.loc[m, f'd{col}'] = final.loc[m, col] - ref[col]
    return final


def plot_response(final, tag, suptitle):
    """Three diagnostics vs lake-level perturbation, each relative to the
    baseline's zero-offset run. Columns are calving laws.
    Colour = sliding law, marker = corrugation direction."""
    quantities = [('dL',       'Δ terminus position (m)'),
                  ('du_t',     'Δ velocity (m/yr)'),
                  ('dgl_flux', 'Δ ice flux (m²/yr)')]
    directions_present = sorted(final['direction'].unique().tolist(),
                                key=direction_order.index)
    configs = [(sl, d) for sl in sliding_laws for d in directions_present
               if len(final[(final['sliding_law'] == sl) &
                            (final['direction'] == d)])]

    fig, axes = plt.subplots(len(quantities), len(calving_laws),
                             figsize=(11, 7.5), sharex=True, squeeze=False)
    for r, (qty, ylabel) in enumerate(quantities):
        for c, cl in enumerate(calving_laws):
            ax = axes[r, c]
            for sl, direction in configs:
                sub = final[(final['sliding_law'] == sl) &
                            (final['direction'] == direction) &
                            (final['calving_law'] == cl)].sort_values('waterline_offset')
                if len(sub) == 0:
                    continue
                st = curve_style(sl, direction)
                x = sub['waterline_offset'].to_numpy(dtype=float)
                ax.plot(x, sub[qty].to_numpy(dtype=float),
                        ms=curve_marker_size, lw=curve_line_width,
                        label=curve_label(sl, direction), **st)
            ax.axvline(0, color='k', lw=0.5); ax.axhline(0, color='k', lw=0.5)
            ax.grid(True, alpha=0.3)
            if r == 0:
                ax.set_title(calving_labels[cl], fontsize=9)
            if r == len(quantities) - 1:
                ax.set_xlabel('lake-level perturbation (m)')
            if c == 0:
                ax.set_ylabel(ylabel, fontsize=9)

    handles = [Line2D([], [], color=sliding_colors[sl], marker='o', ms=6,
                      lw=0, label=sliding_labels[sl])
               for sl in sliding_laws]
    dirs = [d for d in direction_order if d != 'none'
            and any(d == dd for _, dd in configs)]
    handles += [Line2D([], [], color='0.35', marker=direction_markers[d], ms=7,
                       lw=0, label=f'{d} slope')
                for d in dirs]
    axes[0, 0].legend(handles=handles, loc='best', fontsize=7,
                      handletextpad=0.3, borderpad=0.3, labelspacing=0.3)

    fig.tight_layout()
    fig.savefig(f'{figures_path}/{tag}_response.png', dpi=120, bbox_inches='tight')
    plt.show()

    zero = final[np.isclose(final['waterline_offset'], 0.0)]
    if len(zero):
        print('zero-offset baselines:')
        for _, row in zero.sort_values(['calving_law', 'sliding_law',
                                        'direction']).iterrows():
            name = curve_label(row['sliding_law'], row['direction'])
            print(f"  {calving_labels[row['calving_law']]:<13s} {name:<28s} "
                  f"L = {row['L']:8.1f} m   "
                  f"velocity = {row['u_t']:8.2f} m/yr   "
                  f"flux = {row['gl_flux']:10.1f} m²/yr")
```

We now build the six baselines used throughout the paper — the four idealized configurations (two sliding laws × two corrugation directions) and the two Luggye configurations (two sliding laws) — then produce Figure 1.

```python
# Idealized bed corrugation and initial geometry. Head at x=0, terminus at x=length.
# Prograde bed dips to -water_depth at the terminus; corrugation is applied so
# that d(bed)/dx = 0 at the terminus (level plateau) while the terminal water
# depth is preserved. Both corrugation directions are built.
geom_X = np.linspace(0.0, idealized_length, idealized_n_cells + 1)
_bed_prograde = -idealized_water_depth + idealized_bed_slope * (idealized_length - geom_X)

_A       = idealized_bed_corrugation_amplitude
_lambda  = idealized_bed_corrugation_wavelength
_A_min   = idealized_bed_slope * _lambda / (2 * np.pi)
assert _A >= _A_min, (
    f'Corrugation too gentle for a terminus plateau: A = {_A:.2f} m < '
    f'A_min = slope·λ/(2π) = {_A_min:.2f} m. '
    f'Increase idealized_bed_corrugation_amplitude, or reduce λ or the '
    f'prograde slope.')
_k = idealized_bed_slope * _lambda / (2 * np.pi * _A)
_phase_magnitude = float(np.arcsin(_k))

idealized_directions = ['retrograde', 'prograde']
idealized_configs    = [(sl, d) for sl in sliding_laws for d in idealized_directions]

geom_bed = {}
geom_thickness = {}
for direction in idealized_directions:
    sign  = {'retrograde': +1, 'prograde': -1}[direction]
    phase = sign * _phase_magnitude
    _s    = idealized_length - geom_X
    _offset = sign * _A * (np.cos(2 * np.pi * _s / _lambda + phase) - np.cos(phase))
    _bed_profile   = _bed_prograde + _offset
    # H chosen so bed + H = _bed_prograde + idealized_thickness everywhere; the
    # spin-up delivers a smooth initial surface with H thicker over troughs.
    _thick_profile = _bed_prograde + idealized_thickness - _bed_profile
    for sl in sliding_laws:
        geom_bed[(sl, direction)]       = _bed_profile.copy()
        geom_thickness[(sl, direction)] = _thick_profile.copy()

length    = idealized_length
n_cells   = idealized_n_cells
H_initial = Constant(idealized_thickness)
```

```python
# Idealized spin-ups (Section 2.4.1). For each (sliding_law, direction) config,
# spin up under the trivial calving law, then calibrate k (Eq. 3) and f* (Eq. 4).
spinup_kwargs = dict(max_time=spinup_max_time,
                     Δt_init=spinup_Δt_init,
                     Δt_final=spinup_Δt_final,
                     Δt_ramp=spinup_Δt_ramp,
                     tol_V_frac=spinup_tol,
                     tol_L_frac=spinup_tol)

seeds = {}
bed0s = {}
mesh0s = {}
depth_rate = {}
f_star = {}
baselines = {}

for sl, direction in idealized_configs:

    print(f'\n{sliding_labels[sl]} — {direction}')

    fresh = make_initial_state(sl, direction=direction)
    mesh0, bed0 = prepare_extended_bed(fresh, sl, direction=direction)
    bed0s[(sl, direction)] = bed0
    mesh0s[(sl, direction)] = mesh0

    spin_key = fingerprint(
        kind='spinup', sliding_law=sl, direction=direction,
        length=length, n_cells=n_cells,
        idealized_thickness=idealized_thickness,
        idealized_bed_slope=idealized_bed_slope,
        idealized_water_depth=idealized_water_depth,
        corrugation_A=idealized_bed_corrugation_amplitude,
        corrugation_lambda=idealized_bed_corrugation_wavelength,
        spinup_max_time=spinup_max_time, spinup_Δt_init=spinup_Δt_init,
        spinup_Δt_final=spinup_Δt_final, spinup_Δt_ramp=spinup_Δt_ramp,
        spinup_tol=spinup_tol,
        **physics_parts())

    hit = cache_hit('spinup', spin_key, 'npz',
                    force=force_for(force_rerun_spinups, 'idealized'))
    seed = load_state_cache(hit, sl, direction) if hit else None
    if seed is not None:
        print(f'    [{sl[:8]}/{direction[:4]}] loaded cached spin-up '
              f'({os.path.basename(hit)})')
    else:
        seed = run_spinup(
            fresh, bed0, sliding_law=sl, calving_law='trivial',
            desc=f'trivial/{sl[:8]}/{direction[:4]}',
            L_domain=float(mesh0.length),
            **spinup_kwargs,
        )
        save_state_cache('spinup', spin_key, seed)

    seeds[(sl, direction)] = seed

    k = calibrate_depth_rate(seed, baseline_waterline)
    f = calibrate_f_star(seed, baseline_waterline)

    depth_rate[(sl, direction)] = k if k is not None else per_sliding(depth_rate_default, sl)
    f_star[(sl, direction)]     = f if f is not None else per_sliding(f_star_default, sl)

    d = diagnostics(seed)
    print(f'  steady state:  L = {d["L"]:.0f} m,  u_t = {d["u_t"]:.2f} m/yr,  d_t = {d["d_t"]:.1f} m')
    print(f'  calibrated :  k = {depth_rate[(sl, direction)]:.4f} /yr (Eq. 3),  '
          f'f* = {f_star[(sl, direction)]:.4f} (Eq. 4)')

    for cl in calving_laws:
        baselines[(sl, direction, cl)] = {
            'state': seed, 'bed_0': bed0, 'mesh_0': mesh0,
            'depth_rate': depth_rate[(sl, direction)],
            'f_star': f_star[(sl, direction)],
        }

# Stable alias for the combined geometry figure (which needs both baseline sets).
idealized_baselines = baselines
```

```python
# Baseline geometry profile helpers, used by the combined all-experiments figure.
def _sample_profile(entry):
    """Sample one baseline entry: ice surface/base + depth-averaged velocity
    on the experiment mesh, plus the extended bed on the spin-up mesh."""
    state, bed_0, mesh_0 = entry['state'], entry['bed_0'], entry['mesh_0']
    mesh_1 = state['mesh']
    H, s, u = state['H'], state['s'], state['u']
    base = firedrake.project(s - H, s.function_space())

    x1 = np.asarray(mesh_1.X, dtype=float)
    o1 = np.argsort(x1)
    s_a    = np.array(icepack.depth_average(s).at(x1, tolerance=1e-10))[o1]
    base_a = np.array(icepack.depth_average(base).at(x1, tolerance=1e-10))[o1]
    u_a    = np.array(icepack.depth_average(u).at(x1, tolerance=1e-10))[o1]
    x1 = x1[o1]

    x0 = np.asarray(mesh_0.X, dtype=float)
    o0 = np.argsort(x0)
    bed_a = np.array(icepack.depth_average(bed_0).at(x0, tolerance=1e-10))[o0]
    x0 = x0[o0]

    L_ice = terminus_x(state)
    x_hi = L_ice + min(plot_seaward_pad_frac * L_ice, float(x0.max()) - L_ice)
    return dict(x1=x1, s=s_a, base=base_a, u=u_a, x0=x0, bed0=bed_a,
                L=L_ice, x_hi=x_hi)


def _draw_profile(ax, p, y_lo):
    ax.fill_between(p['x0'], y_lo, p['bed0'],
                    color=profile_colors['bedrock'], label='rock')
    ax.fill_between(p['x0'], p['bed0'], baseline_waterline,
                    where=(p['bed0'] < baseline_waterline),
                    color=profile_colors['water'], label='water')
    ax.fill_between(p['x1'], p['base'], p['s'],
                    color=profile_colors['ice'], label='ice')
    ax.plot(p['x1'], p['s'],    color=profile_colors['surface'],  lw=1.1)
    ax.plot(p['x1'], p['base'], color=profile_colors['ice_edge'], lw=1.1)
    ax.plot(p['x0'], p['bed0'], color=profile_colors['bed_edge'], lw=1.2,
            label='bed')


def _profile_legend(ax, fontsize=6):
    h, l = ax.get_legend_handles_labels()
    h.append(Line2D([], [], color=profile_colors['velocity'], lw=1.0))
    l.append('u')
    ax.legend(h, l, fontsize=fontsize, loc='upper right')


def plot_combined_baseline_profiles(sets, tag='all',
                                    title='Steady-state geometries for all experiments'):
    """Rows are sliding laws; columns are experiments. Each SET (idealized,
    Luggye) shares its x/elevation/velocity scales across its own columns but
    not across sets (the two differ by ~2.5x in length and ~3x in speed)."""
    columns = []
    for si, (bl, specs) in enumerate(sets):
        for col_title, direction in specs:
            columns.append(dict(title=col_title, set_idx=si,
                                direction=direction, baselines=bl))

    nrow, ncol = len(sliding_laws), len(columns)
    fig, axes = plt.subplots(nrow, ncol, figsize=(4.4 * ncol, 3.0 * nrow),
                             squeeze=False)

    profiles = {}
    for c, col in enumerate(columns):
        for r, sl in enumerate(sliding_laws):
            key = (sl, col['direction'], calving_laws[0])
            if key in col['baselines']:
                profiles[(r, c)] = _sample_profile(col['baselines'][key])

    lims = {}
    for si in sorted({col['set_idx'] for col in columns}):
        cols_i = [c for c, col in enumerate(columns) if col['set_idx'] == si]
        ps = [profiles[(r, c)] for c in cols_i for r in range(nrow)
              if (r, c) in profiles]
        if not ps:
            continue
        x_hi = max(p['x_hi'] for p in ps)
        u_max = max((float(np.nanmax(np.abs(p['u']))) for p in ps if len(p['u'])),
                    default=0.0)
        lims[si] = dict(
            x=(min(float(p['x0'].min()) for p in ps), x_hi),
            y=(min(min(float(p['bed0'][p['x0'] <= x_hi].min()) for p in ps),
                   baseline_waterline) - 20,
               max(float(p['s'].max()) for p in ps) + 20),
            u=(0.0, u_max * 1.05 if u_max > 0 else 1.0),
        )
    first_of_set = {si: min(c for c, col in enumerate(columns) if col['set_idx'] == si)
                    for si in lims}
    last_of_set = {si: max(c for c, col in enumerate(columns) if col['set_idx'] == si)
                   for si in lims}

    for c, col in enumerate(columns):
        si = col['set_idx']
        for r, sl in enumerate(sliding_laws):
            ax = axes[r, c]
            if (r, c) not in profiles:
                ax.axis('off')
                continue
            p = profiles[(r, c)]
            _draw_profile(ax, p, lims[si]['y'][0])
            ax.set_xlim(*lims[si]['x']); ax.set_ylim(*lims[si]['y'])

            ax2 = ax.twinx()
            ax2.plot(p['x1'], p['u'], color=profile_colors['velocity'], lw=1.0)
            ax2.set_ylim(*lims[si]['u'])
            if c == last_of_set[si]:
                ax2.tick_params(axis='y', labelcolor=profile_colors['velocity'],
                                labelsize=7)
                ax2.set_ylabel('u (m/yr)',
                               color=profile_colors['velocity'], fontsize=8)
            else:
                ax2.tick_params(axis='y', labelright=False, right=False)

            if c == first_of_set[si]:
                ax.set_ylabel('elevation (m)', fontsize=8)
            else:
                ax.tick_params(axis='y', labelleft=False)
            if r == 0:
                ax.set_title(col['title'], fontsize=11)
            if r == nrow - 1:
                ax.set_xlabel('x (m)', fontsize=8)
            else:
                ax.tick_params(axis='x', labelbottom=False)
            ax.tick_params(labelsize=7)

    for r, sl in enumerate(sliding_laws):
        axes[r, 0].annotate(sliding_labels[sl], xy=(-0.28, 0.5),
                            xycoords='axes fraction', rotation=90,
                            va='center', ha='center', fontsize=10)
    _profile_legend(axes[0, 0], fontsize=6)

    fig.tight_layout(rect=(0.02, 0, 1, 0.97))
    fig.savefig(f'{figures_path}/{tag}_combined_profiles.png', dpi=150,
                bbox_inches='tight')
    plt.show()
```

## 3 Results

Results are shown in Figures 2 and 3; all diagnostics are reported relative to the zero-perturbation run of the same configuration. We forward-model each perturbed configuration for a single year. Over this interval the geometry cannot adjust appreciably, so each experiment isolates the short-term response of the ice before terminus position, thickness, and lake level begin to feed back on one another. By construction the idealized configurations sit in steady state at zero perturbation, and the Luggye configurations, though not in steady state, predict zero net terminus migration at $t=0$ (Section 2.4).

We first hold the calving law fixed and ask how the sliding law shapes the response. Under the rate and trivial laws the two behave alike, because both tie the calving rate to the flow rate (Section 4.1), and the sliding law sets both the magnitude and the sign of the response. With power-law sliding (Equation 1), a rising lake slows the terminus and reduces the flux delivered to it (by up to $\sim$8 m yr$^{-1}$ and $\sim$300 m$^3$ yr$^{-1}$ per unit width at a 10 m rise) and a falling lake does the reverse. With effective-pressure sliding (Equation 2) the response is weaker and no longer monotonic: on the prograde bed, velocity and flux rise for a perturbation of *either* sign, tracing the parabolic curves of Figure 2 (we discuss the underlying mechanism in Section 4.1). The terminus itself retreats under a rising lake for both sliding laws (more steeply under power-law sliding) in the rate law, and is held fixed by construction under the trivial law.

The position law (Equation 4) breaks this pattern. The front is now placed kinematically, so flow no longer feeds back on calving, and the response turns strongly asymmetric: a rising lake drives large changes (retreat of up to $\sim$100 m, with velocity and flux rising by $\sim$18 m yr$^{-1}$ and $\sim$800 m$^3$ yr$^{-1}$ per unit width at a 10 m rise) while a falling lake does comparatively little. With the calving law fixed at the position law, the sliding law now matters little; the two sliding laws produce nearly identical curves (Figure 2, centre column).

## 4 Discussion and conclusions

### 4.1 Idealized flowline model

Under the rate and trivial laws, the flow rate influences the calving rate, which is why the two behave alike. For the rate law this coupling is incidental: faster flow moves the front more quickly, altering the terminal water depth and hence the calving rate through Equation 3. For the trivial law it is imposed directly by Equation 5. The position law suppresses the coupling: however fast the front tries to advance, it must sit wherever the calving criterion is met at each timestep. The contrast between bed and sliding law is therefore far less pronounced under the position law than under the other two.

The two sliding laws differ, in turn, because they admit different subsets of the mechanisms in Section 1. A rising lake always raises the backpressure on the terminal face, reducing extensional stress and slowing the glacier; this acts under both laws. Under power-law sliding it is the only direct route by which lake level enters the momentum balance, so a rising lake unambiguously reduces the flux delivered to the lake. Under effective-pressure sliding a second, opposing mechanism appears: a rising lake lowers the fraction above flotation near the front, weakening basal resistance and speeding the glacier up. The net response is the residual of the two, which is why, for the rate and trivial laws, the effective-pressure curves are smaller than the power-law ones and can rise for perturbations of either sign.

Bed geometry matters only once the sliding law couples it to basal resistance. Under power-law sliding the bed enters only through ice thickness, so our prograde and retrograde configurations, which share their broad geometry, behave almost identically for a given calving law. Under effective-pressure sliding the same small geometric difference changes the basal resistance near the front, and the two beds diverge. Bed slope, in short, matters to the extent that the sliding law lets it.

These results point to distinct feedbacks between glacier dynamics and lake level, depending on the governing equations. Under the rate and trivial laws, a rising lake increases discharge when basal resistance depends on water pressure (effective-pressure sliding), which would in turn raise the lake further; a pure power-law relation instead reduces discharge, a self-limiting negative feedback. The sign of the feedback is thus set jointly by the sliding law, the calving law, and the bed geometry. Because our experiments capture only the short-term response, they fix the initial sign of these feedbacks rather than their eventual outcome; resolving that outcome would require lake-volume estimates, which are meaningful only in a plan-view model and lie beyond our scope here.

### 4.2 Luggye flowline model

Luggye's inverted beds are slightly retrograde at the terminus (see the Luggye panels of Figure 1), and under the rate and trivial laws it behaves accordingly: power-law sliding ties velocity and flux to lake level, effective-pressure sliding barely responds. The idealized experiments, therefore, anticipate the observation-informed ones well. The position law is the exception. Here the largest responses of the study appear under *power-law* sliding, and on the lake-lowering side: with the front placed by proximity to flotation, a drawdown lets the terminus advance far down the deep, retrograde bed returned by the power-law inversion ($\sim$97 m water depth), and velocity and flux follow. The effective-pressure inversion places the terminus in much shallower water ($\sim$42 m) and produces a milder response. We attribute the contrast to bed geometry acting through the kinematic position criterion, rather than to the sliding mechanism itself: the pure power-law relation carries no basal-resistance feedback, yet it drives the extremes.

We also note that the two Luggye inversions are not equally credible. Because each bed is recovered under an assumed sliding law, the two place the terminus in substantially different water depths. The field-observed depth of $\sim$50 m at the Luggye terminus aligns with the shallower, effective-pressure inversion ($\sim$42 m) rather than the power-law one ($\sim$97 m). We therefore take the effective-pressure results as the more likely representation of Luggye's behaviour, and infer that its calving flux is comparatively insensitive to modest changes in lake level, provided that terminus migration is rate-limited rather than position-limited.

### 4.3 Limitations

This study, of course, has made several simplifying assumptions, which we discuss here.

#### 4.3.1 The role of surface mass balance

In order to focus on the ice dynamical processes, we have neglected surface mass balance, both in the idealized model and the Luggye model. We consider this the most reasonable approach, especially given that we lack any measurements of Luggye's true surface mass balance. Nevertheless, it should be noted that the bed profiles obtained for Luggye would likely differ had we accounted for additional surface-lowering processes (van Pelt et al., 2013). As the bed location ultimately determines the ice thickness, and thereby the velocity solutions and grounding line ice flux baselines, results for the Luggye model should be taken as indicative of broad relationships between parameters, and not as high-confidence predictions.

#### 4.3.2 Choice and calibration of sliding and calving laws

Many more variations of sliding and calving laws exist than we have considered here. Our goal, however, is to explore the qualitative differences that emerge from contrasting treatments of basal resistance and terminus migration, rather than to exhaustively survey the available laws. Equations 1 and 2 are chosen as mutually compatible end members. Meanwhile, the calving law choices of Equations 3 and 4 allow us to isolate the effects of the rate-vs.-position formulation.

Calibration of the calving laws in particular may also influence results. For the idealized experiments these choices have been made naturally, and such that the resulting steady-state configurations remain in steady state regardless of the calving law chosen. In contrast, our Luggye setup is not in steady state, and the choice to calibrate each calving law to minimize terminus retreat at $t = 0$ is admittedly arbitrary. Unfortunately, the difficulty of accurate calibration is a known obstacle in the development of high-quality, predictive calving laws (e.g., Amaral et al., 2020; Choi et al., 2018).

#### 4.3.3 Flowline geometry

In our flowline models, the gravitational driving stress is balanced by basal drag and longitudinal stress gradients alone. In a natural glacier setting, lateral drag provides some resistance. A plan-view model would likely yield more detailed insights than the flowline models considered here, and we consider the development of such models a natural next step in our analysis.

### 4.4 Conclusions

The dynamic response of a lake-terminating glacier to a change in lake level is not set by the lake alone. A clear hierarchy emerges. The calving law decides whether the ice feels the lake at all: where the terminus is placed by a competition between flow and calving (the rate and trivial laws), the sliding law governs the response; where it is placed kinematically (the position law), that dependence is largely suppressed. Given a calving law under which the sliding law matters, the sliding law then fixes the sign — whether calving flux falls, holds steady, or rises as the lake deepens — and, where basal resistance is tied to water pressure, the local bed slope decides which occurs. In every case, letting basal resistance weaken with water pressure amplifies the response to a deepening lake.

For Luggye, the sliding law most consistent with the observed $\sim$50 m terminus water depth implies that calving flux is comparatively insensitive to modest lake-level change, provided the front is free to migrate at a rate set by flow and calving. Positioned instead by its proximity to flotation, the same glacier becomes far more sensitive. Constraining the ice-dynamical component of GLOF risk will therefore mean pinning down basal conditions and terminus-migration style together; neither can be inferred from lake level, or from the other, in isolation.
